```python
# 1. Define your DNA sequence (The 'Reagent')
dna_sequence = "ATGGCCATTGTAATGGGCCGCTGAAAGGGTGCCCGATAG"

# 2. Transcription: DNA -> RNA
rna_sequence = dna_sequence.replace("T", "U")
print(f"RNA Sequence: {rna_sequence}")

# 3. Translation Table (The 'Codon Map')
codon_map = {
    'AUG': 'M', 'GCC': 'A', 'AUU': 'I', 'GUA': 'V',
    'GGG': 'G', 'CCG': 'P', 'CUG': 'L', 'AAA': 'K',
    'GGU': 'G', 'CCC': 'P', 'GAU': 'D', 'UAG': '*' # * is a Stop Codon
}

# 4. Translation Process
protein = ""
for i in range(0, len(rna_sequence), 3):
    codon = rna_sequence[i:i+3]
    # Check if the codon is in our map (to avoid errors)
    amino_acid = codon_map.get(codon, "?")
    protein += amino_acid

print(f"Translated Protein: {protein}")
```
